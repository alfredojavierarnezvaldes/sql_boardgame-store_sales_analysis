# Sales Analysis Report

This notebook presents an analysis of a simulated online boardgame store database corresponding to the month of october 2025, using SQL queries.
The goal is to extract meaningful business insights related to sales, customers, and payments.




## Dataset Description

The database simulates an online store and contains the following tables:

- Customers
- Products
- Categories
- Orders
- Order Items
- Payments

The data was created manually for learning and portfolio purposes.





## Analysis Scope

The analysis focuses on:
- Sales performance
- Customer behavior
- Revenue distribution by department
- Payment method usage



## Business Question 1
Which cities generate the highest revenue?

SELECT 
    c.department,
    SUM(oi.quantity * oi.price) AS total_revenue
FROM customers c
JOIN orders o ON c.customer_id = o.customer_id
JOIN order_items oi ON o.order_id = oi.order_id
GROUP BY c.department
ORDER BY total_revenue DESC;

Cochabamba generates the highest revenue, with 20,850, followed by Santa Cruz with 9,500. La Paz, Oruro, and Sucre generate considerably lower revenue. This indicates that sales are strongly concentrated in Cochabamba, which represents the main revenue-generating market during the analyzed period.




## Business Question 2
Who are the top customers by total spending?

SELECT 
    c.full_name,
    SUM(oi.quantity * oi.price) AS total_spent
FROM customers c
JOIN orders o ON c.customer_id = o.customer_id
JOIN order_items oi ON o.order_id = oi.order_id
GROUP BY c.full_name
ORDER BY total_spent DESC
LIMIT 5;

Carla Butrón is the highest-spending customer, with a total spending of 5,300. The remaining customers in the top five show lower but relatively close spending levels, ranging from 2,000 to 2,700.




## Business Question 3
What payment methods are most frequently used?

SELECT 
    payment_method,
    COUNT(*) AS total_transactions
FROM payments
GROUP BY payment_method
ORDER BY total_transactions DESC;

QR is the most frequently used payment method, with 11 transactions, followed closely by debit card with 10 transactions. Cash is less frequently used, with 4 transactions.




## Final Conclusions

- Cochabamba has the strongly highest revenue during the analyzed period.
- Carla Butrón is the highest-spending customer, with a total spending of 5,300.
- QR is the most frequently used payment method, followed closely by debit card.
- SQL queries combined with structured analysis allow for actionable business insights.
